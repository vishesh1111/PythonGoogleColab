## File Handling



### Some Theory

##### Types of data used for I/O:
- Text - '12345' as a sequence of unicode chars
- Binary - 12345 as a sequence of bytes of its binary equivalent

##### Hence there are 2 file types to deal with
- Text files - All program files are text files
- Binary Files - Images,music,video,exe files

### How File I/O is done in most programming languages

- Open a file
- Read/Write data
- Close the file

```python
# f.close is used for handling large GB files as they are loaded in the memory and if not closed they will remain in the memory and will consume more memory(RAM)
# also anybody can access the files as they are in the RAM --> Security Reason
```

```python
# case 1 - if the file is not present
f = open('sample.txt','w')
f.write('Hello world')
f.close()
# since file is closed hence this will not work
f.write('Hello')
```

```python
# write multiline strings
f = open('sample1.txt','w')
f.write('hello world')
f.write('\nhow are you?')
f.close()
```

```python
# case 2 - if the file is already present
f = open('sample.txt','w')
f.write('homie comes')
f.close()
```

```python
# append mode
f = open('/content/sample1.txt','a')
f.write('\nThis will be appended')
f.close()
```

```python
# write multiple lines
L = ['hello\n','hi\n','how are you\n','I am fine']

f = open('/content/temp/sample.txt','w')
f.writelines(L)
f.close()
```

```python
# reading from files

# -> using read()
f = open('/content/sample.txt','r')
s = f.read()
print(s)
f.close()



# reading upto n chars
f = open('/content/sample.txt','r')
s = f.read(10)
print(s)
f.close()



# readline() -> to read line by line. --> don't load everything at once
f = open('/content/sample.txt','r')
print(f.readline(),end='')
print(f.readline(),end='')
f.close()
# readline changes the line and also the print changes line , so we use the (end)



# reading entire using readline
f = open('/content/sample.txt','r')
while True:

  data = f.readline()

  if data == '':  # data equal to empty
    break
  else:
    print(data,end='')

f.close()
```

### Using Context Manager (With)
1. It's a good idea to close a file after usage as it will free up the resources
2. If we dont close it, garbage collector would close it
3. with keyword closes the file as soon as the usage is over


```python
# with
with open('/content/sample1.txt','w') as f:
  f.write('written using with')


# try f.read() now
with open('/content/sample.txt','r') as f:
  print(f.readline())




# moving within a file ---> 10 char then 10 char
with open('sample.txt','r') as f:
  print(f.read(10))   # first 10 char
  print(f.read(10))   # next 10 char
  print(f.read(10))   # so on...
  print(f.read(10))
```

```python
# seek and tell function:-
# seek --> used to move the file pointer (cursor) to a specific position in a file.       EX:- YT VIDEO Timeline(red bar)
# tell --> gives the current character or where the buffer is currently now or tells what next character will be processing

with open('sample.txt','r') as f:
  f.seek(15)
  print(f.read(10))
  print(f.tell())

  print(f.read(10))
  print(f.tell())
```

```python
# seek during write
with open('sample.txt', 'w') as f:
  f.write('Hello')
  f.seek(0)
  f.write('X')
```

## Serialization:- process of converting python datatypes to JSON Format
##De-Serialization:- process of converting JSON to python dataTYPES

```python
import json
lst = [1,2,3,4,5]
with open('demo.json','w') as f:
  json.dump(lst,f)
```

```python
# easily type conversion is done
import json
with open('demo.json','r') as f:
  print(json.load(f))


# incase of tuple:- it will be stored in form of LIST!!!
t = (1,2,3,4,5,6)
with open('demo.json','w') as f:
  json.dump(t,f)
```

```python
# Serialization & Deserialization with custom objects:-
class Person:
  def __init__(self,fname,lname,age,gender):
    self.fname = fname
    self.lname = lname
    self.age = age
    self.gender = gender

person = Person('Vishesh','verma', 19,'male')
```

```python
import json
def show_object(person):
  if isinstance(person,Person):
    return "{} {} age  -> {}".format(person.fname,person.lname,person.age,person.gender)


with open('demo.json','w') as f:
  json.dump(person,f,default=show_object)
```

### Pickling
`Pickling` is the process whereby a Python object hierarchy is converted into a byte stream, and `unpickling` is the inverse operation, whereby a byte stream (from a binary file or bytes-like object) is converted back into an object hierarchy.

Pickle lets the user to store data in binary format. JSON lets the user store data in a human-readable text format.

```python
 class Person:

  def __init__(self,name,age):
    self.name = name
    self.age = age

  def display_info(self):
    print('Hi my name is',self.name,'and I am ',self.age,'years old')
```

```python
p = Person('Vishesh',19)
```

```python
# pickle dump
import pickle
with open('person.pkl','wb') as f:
  pickle.dump(p,f)
```
