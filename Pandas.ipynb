# Pandas Series


```python
# Pandas Series = is like a column in a table , it is a 1D array holding data of any type

import pandas as pd
import numpy as np

# Creating pandas series from LISTS

#String:-

country = ['INDIA','USA', 'CHINA','RUSSIA']
pd.Series(country)            # if bydefault the dtype = object then we can say that it is string datatype
```

```python
# From Integers:-
runs = [13,24,56,78,200]
pd.Series(runs)
```

```python
# By Making them in (Custom Index):-
marks = [67,57,89,100]
subjects = ['EVS', 'DAA', 'DSA','DBMS']
pd.Series(marks, index = subjects, name = 'SEM2') # name is basically used to set name for an series
```

## Pandas Series from DICT


```python
marks = {
    'EVS': 67,
    'DAA': 57,
    'DSA': 89,
    'DBMS': 100
}
marks_series = pd.Series(marks, name = 'NEW DICTIONARY')
marks_series
```

## Series Attributes:-


```python
# Size:- gives number of elements in the series
marks_series.size

#dtype:- gives datatype
marks_series.dtype

# name:- returns the name specified for an series
marks_series.name

# index:- gives all index for the series
marks_series.index

# isunique:- gives true for all unique items in the series
marks_series.is_unique

# values:- returns the all values that are stored in the series
marks_series.values

```

## Filtering a dataFrame based on given condition

```python
# 1. find all the final winners of every year in The IPL
ipl.head(2)
main = ipl['MatchNumber'] == 'Final'
Final = ipl[main]
Final[['Season','WinningTeam']]
```

```python
# one liner code
ipl[ipl['MatchNumber'] == 'Final'][['Season','WinningTeam']]
```

```python
# ques -> how many superover have occured
ipl[ipl['SuperOver'] == 'Y'].shape[0]
```

```python
# que -> how many matches has csk won in kolkata!
ipl[(ipl['City'] ==  'Kolkata') & (ipl['WinningTeam'] == 'Chennai Super Kings')] # using bitwise and operations -> bcz we are performing operations in boolean values
# ipl[(ipl['City'] ==  'Kolkata') & (ipl['WinningTeam'] == 'Chennai Super Kings')].shape[0]
```

```python
# ques -> toss winner is match winner (good/bad) in percentage?
(ipl[ipl['TossWinner'] == ipl['WinningTeam']].shape[0] / ipl.shape[0]) * 100
```

```python
# ques ->> movies with rating higher than 8 and votes>10k
mov[(mov['imdb_rating'] > 8) & (mov['imdb_votes'] > 10000)].shape[0] # 42/1600
(mov[(mov['imdb_rating'] > 8) & (mov['imdb_votes'] > 10000)].shape[0] / mov.shape[0] )* 100
```

```python
# ques -> action movies with rating > 7.5
# mask1 = mov['genres].str.contains('Action')
mask1 = mov['genres'].str.split('|').apply(lambda x: 'Action' in x)
mask2 = mov['imdb_rating'] > 7.5
mov[mask1 & mask2].shape[0] # 33/1629
```
