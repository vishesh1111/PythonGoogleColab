# Pandas Series


```python
# Pandas Series = is like a column in a table , it is a 1D array holding data of any type

import pandas as pd
import numpy as np

# Creating pandas series from LISTS

#String:-

country = ['INDIA','USA', 'CHINA','RUSSIA']
pd.Series(country)            # if bydefault the dtype = object then we can say that it is string datatype
```

```python
# From Integers:-
runs = [13,24,56,78,200]
pd.Series(runs)
```

```python
# By Making them in (Custom Index):-
marks = [67,57,89,100]
subjects = ['EVS', 'DAA', 'DSA','DBMS']
pd.Series(marks, index = subjects, name = 'SEM2') # name is basically used to set name for an series
```

## Pandas Series from DICT


```python
marks = {
    'EVS': 67,
    'DAA': 57,
    'DSA': 89,
    'DBMS': 100
}
marks_series = pd.Series(marks, name = 'NEW DICTIONARY')
marks_series
```

## Series Attributes:-


```python
# Size:- gives number of elements in the series
marks_series.size

#dtype:- gives datatype
marks_series.dtype

# name:- returns the name specified for an series
marks_series.name

# index:- gives all index for the series
marks_series.index

# isunique:- gives true for all unique items in the series
marks_series.is_unique

# values:- returns the all values that are stored in the series
marks_series.values

```

## Series using read_csv


```python
subs = pd.read_csv('/subs.csv').squeeze()
subs
```

```python
# with 2 columns:-
vk = pd.read_csv('/kohli_ipl.csv',index_col = 'match_no').squeeze()
```

```python
# with 2 columns:- squeeze = “Remove extra dimensions if they are unnecessary” = converts -> single-column DataFrame -> Series
movies = pd.read_csv('/bollywood.csv',index_col = 'movie').squeeze()
movies
```

## Series Methods:-


```python
# head = by default it gives starting 5 data from datasets & canbe any number is specifed(UPTO),and tail() = gives last(bottom) 5 data from datasets
subs.head()
subs.tail(10)
```

```python
# sample:- gives one row randomly -> used when the data is somewhat biased
movies.sample()
```

```python
# value_counts ->data me each item ka frequency (returns)
movies.value_counts()
```

```python
#sort_values-> sort values in order(ascending or descending)
vk.sort_values()
vk.sort_values(ascending = False).head(1).values[0] # example of method chaining!!!!  O/P of another method is I/P of another method

# vk.sort_values(inplace=True) inplace gives us changes in the original dataset
```

```python
# sort_index = used to sort index in alphabetical order
movies.sort_index()
```

# Series Maths Method

```python
# count:- used to count total no. of items in series , it doesnt count missing values in the series, similar to size(series attributes)-> but it also counts number of missing values in the series
vk.count()

# sum & Product:-
subs.sum()
subs.prod()


# mean,median,mode,std,var
subs.mean()
subs.median()
subs.mode() # aparticular values the comes more in frequency
subs.std()
subs.var()


# min/max
subs.min()
subs.max()

# describe() gives everything at one place
vk.describe()
```
