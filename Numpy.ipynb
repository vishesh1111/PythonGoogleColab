Broadcasting

```python
# the term brodcasting defines/describes how Numpy treats arrays with diffrent shapes during arithmetic operations
# the smaller array is Broadcast across the larger array so that they have compatible shapes -> smaller arrays bcms larger enough to add

# Brodcasting Rules:-
# 1. make the two arrays have the same number of dimensions -> add new dimension with size 1 to the head of the array with the smaller dimensions
# 2. make each dimension of the two arrays the same size -> if size doesnt match -> dimension with size 1 are streched to the size of the other array
# Ex:- 2D(3,3) + 1D(3) = Broadcast (3,3) + (1,3) = still the size doesnt match -> now streched the size of(1,3) specially 1 by 3 and make it (3,3) -> streched till it is not matched then broadcasting is performed

import numpy as np
a = np.arange(12).reshape(4,3)
b = np.arange(3)

print(a)
print(b)

print(a+b)  #performed

a = np.arange(12).reshape(3,4)
b = np.arange(3)
# print(a+b) throws an error bcz the 1 is headed to 3 i.e (1,3) -> then -> (3,3) -> still didnt match with the inital array so error comes
```

```python
# sigmoid function:- studied in LOGISTCIC REGRESSION ML ALGO
import numpy as np
def sigmoid(array):
  return 1/(1+np.exp(-(array)))

a = np.arange(10)
sigmoid(a)


# 2. Mean square error -> used in linear regression  ALGO i.e formula = (predicted-actual)^2 + summation of each elements error
actual = np.random.randint(1,50,25)
predicted = np.random.randint(1,50,25)
predicted

```

```python
def mse(actual,predicted):
  return np.mean(np.square(actual-predicted))

mse(actual,predicted)
```
