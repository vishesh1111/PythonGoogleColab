```python
# Array Attributes:- Array attributes are properties of a NumPy array that describe its structure (shape, size, type, etc.).-> Like metadata of the array
# 1.ndim:- gives number of dimension from an given array
# 2.shape:- gives rows and columns in each dimension
# 3.size:- gives number of items in the given array
# 4.itemsize:- gives size of the memory used by each item -> In Bytes
# 5.dtype:- gives the datatype of the array with size(i.e 32,64 bytes)
# 6.nbytes:- gives total memory used
# 7.reshape:- helps to change the shape of the array(DIMENSION)
# 8.T:- helps to transpose of the array
```

```python
import numpy as np
a1 = np.arange(10)
a2 = np.arange(12, dtype=float).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2)
```

```python
# Changing the datatype of the exisiting array:-
# astype:-
a3.dtype # by default = 64
a3.astype(np.int32)
```
